```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
# Data
data = {
    'Attendance':[95,80,75,60,90,85,55,70,np.nan,88],
    'Study_Hours':[6,5,4,2,7,6,1,3,5,np.nan],
    'Internal_Marks':[28,24,22,12,30,26,10,18,25,27],
    'Result':['Pass','Pass','Pass','Fail','Pass',
              'Pass','Fail','Fail','Pass','Pass']
}

df = pd.DataFrame(data)

print("Raw Dataset")
print(df)
```

```python
#Sampling means selecting representative data from a larger dataset.
# randomly select 8 records.

sample_df = df.sample(n=8, random_state=1)

print("Sampled Dataset")
print(sample_df)
```

```python
# Data Pre-processing
print(sample_df.isnull().sum())
```

```python
# Data Cleaning
# Replace missing values using the column mean.

sample_df['Attendance'].fillna(sample_df['Attendance'].mean(), inplace=True)

sample_df['Study_Hours'].fillna(sample_df['Study_Hours'].mean(), inplace=True)

print(sample_df)
```

```python
# Encode Target Variable
# Machine learning algorithms require numerical values.

encoder = LabelEncoder()

sample_df['Result'] = encoder.fit_transform(sample_df['Result'])

print(sample_df)
```

```python
# Data Partitioning
# Separate features and target.

X = sample_df[['Attendance',
               'Study_Hours',
               'Internal_Marks']]

y = sample_df['Result']
```

```python
# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.25,
        random_state=42)

print("Training Records :",len(X_train))
print("Testing Records :",len(X_test))

# Training data teaches the model.
# Testing data evaluates the model.
```

```python
# Building the Model

# Create the Linear Regression model

from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

print("Model trained successfully.")
```

```python
# Display the Statistical Model
# Show the regression equation learned by the model.

print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)

print("\nRegression Equation:")
print(f"Internal Marks = {model.intercept_:.2f} + ({model.coef_[0]:.2f} × Study Hours)")
```

```python
# Predict Internal Marks

y_pred = model.predict(X_test)

print("Actual Marks")
print(y_test.values)

print("\nPredicted Marks")
print(y_pred)
```

```python
# Evaluate the Model

from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("Mean Absolute Error:", mae)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)

print("Mean Squared Error:", mse)
```

```python
import numpy as np

rmse = np.sqrt(mse)

print("Root Mean Squared Error:", rmse)
```

```python
# R² Score → Indicates how well the regression model explains the variation in marks (closer to 1 is better).
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R2 Score:", r2)
```

```python
# Predict a New Student's Marks

new_student = [[7]]      # Study Hours = 7

predicted_marks = model.predict(new_student)

print("Predicted Internal Marks =", predicted_marks[0])
```
