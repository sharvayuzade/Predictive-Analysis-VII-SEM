```python
import numpy as np
```

```python
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([10, 20, 30, 40, 50, 60])

print("Input values:", X.ravel())
print("Target values:", y)
```

```python

new_value = 3.5
k = 2
```

```python

distances = abs(X.ravel() - new_value)
nearest = np.argsort(distances)[:k]

print("Nearest input values:", X[nearest].ravel())
```

```python

prediction = np.mean(y[nearest])

print("Prediction:", prediction)
```
