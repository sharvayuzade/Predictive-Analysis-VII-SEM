```python
from sklearn.metrics import mean_absolute_percentage_error
import math
import numpy as np
```

```python
y = np.array([50,60,70,80,90])
print(y)
yh = np.array([45,63,68,85,87])
print(yh)
```

```python
mape = mean_absolute_percentage_error(y,yh)
print(mape*100)
```

```python
y2 = np.array(y)
y3 = np.array(yh)
```

```python
mspe = np.mean(((y2 - y3) / y2) ** 2)
print(mspe*100)
```

```python
rmspe = np.sqrt(mspe)
print(rmspe*100)
```
