```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
x=np.array([100,150,200,250,300])
y=np.array([20,25,30,35,40])
x_mean = np.mean(x)
y_mean = np.mean(y)
print("x mean:", x_mean)
print("y mean:",y_mean)
```

```python
b1=np.sum((x-x_mean)*(y-y_mean))/np.sum((x-x_mean)**2)
print("slope b1:",b1)
```

```python
b0=y_mean-b1*x_mean
print("intercept b0:",b0)
```

```python
y_predicted = b0 + b1*x
print("y predicted values:", y_predicted)
```

```python
MAE = mean_absolute_error(y, y_predicted)
print("MAE:",MAE)
```

```python
MSE = mean_squared_error(y, y_predicted)
print("MSE:",MSE)
```

```python
RMSE = np.sqrt(MSE)
print("RMSE:",RMSE)
```
