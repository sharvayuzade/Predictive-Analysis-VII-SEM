```python
import math
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_absolute_percentage_error
```

```python
x = np.array([1,2,3,4,5])
y = np.array([20,40,50,65,80])
print(x,y)
```

```python
x_mean = x.mean()
y_mean = y.mean()
print(x_mean,y_mean)
```

```python
#design matrix
xm = np.array([[1, x_i, x_i**2] for x_i in x])
print(xm)
```

```python
x_transpose = xm.T
x_transpose_x = np.dot(x_transpose, xm)
print(x_transpose)
print()
print(x_transpose_x)
```

```python
x_transpose_y = np.dot(x_transpose,y)
print(x_transpose_y)
```

```python
x_transpose_x_inv = np.linalg.inv(x_transpose_x)
print(x_transpose_x_inv)
```

```python
theeta = np.dot(x_transpose_x_inv, x_transpose_y)
print(theeta)
```

```python
#print euqation
print(f"y = {theeta[0]} + {theeta[1]}*)x+{theeta[2]}*x^2")
```

```python
#residual error calculation
y_pred = np.dot(xm, theeta)
residuals = y - y_pred
print(residuals)
```

```python
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = math.sqrt(mse)
r2 = r2_score(y, y_pred)
mape = mean_absolute_percentage_error(y, y_pred)
mspe = np.mean((residuals / y) ** 2)
rmspe = np.sqrt(mspe)
```

```python
print("MSE:", mse)
print("RMSE:", rmse)
print("MAE:", mae)
print("R2:", r2)
print("MAPE:", mape)
print("MSPE:", mspe)
print("RMSPE:", rmspe)
```
